```python
# Merge swap events with stations to get the station grid tariff rate
df7_merged = df7.merge(df5[['station_id', 'grid_tariff_inr_kwh']], on='station_id', how='left')

# Calculate recharging electricity cost for each event
df7_merged['recharge_cost_inr'] = df7_merged['energy_to_recharge_kwh'] * df7_merged['grid_tariff_inr_kwh']

# Calculate daily trends with contribution margins included
daily_performance = df7_merged.groupby('date').agg(
    total_attempts=('event_id', 'count'),
    completed_swaps=('event_type', lambda x: (x == 'swap_completed').sum()),
    failed_swaps=('event_type', lambda x: x.isin(['failed_no_charged_battery', 'failed_system_error']).sum()),
    revenue_inr=('amount_charged_inr', 'sum'),
    total_cost_inr=('recharge_cost_inr', 'sum')
).reset_index()

# Compute key daily metrics
daily_performance['failure_rate'] = daily_performance['failed_swaps'] / daily_performance['total_attempts']
daily_performance['contribution_margin_inr'] = daily_performance['revenue_inr'] - daily_performance['total_cost_inr']
daily_performance['avg_margin_per_completed_swap'] = daily_performance['contribution_margin_inr'] / daily_performance['completed_swaps']

# Preview the calculated metrics
display(daily_performance.head())

# Let's plot the daily trends to see how these metrics behave over time
import matplotlib.pyplot as plt

fig, axes = plt.subplots(3, 1, figsize=(14, 18), sharex=True)

# Subplot 1: Volumes
axes[0].plot(daily_performance['date'], daily_performance['completed_swaps'], label='Completed Swaps', color='green')
axes[0].set_title('Daily Completed Swaps Over Time')
axes[0].set_ylabel('Swap Volume')
axes[0].grid(True)

# Subplot 2: Financial Metrics
axes[1].plot(daily_performance['date'], daily_performance['revenue_inr'], label='Revenue', color='blue')
axes[1].plot(daily_performance['date'], daily_performance['contribution_margin_inr'], label='Contribution Margin', color='orange')
axes[1].set_title('Daily Revenue vs Contribution Margin (INR) Over Time')
axes[1].set_ylabel('Amount (INR)')
axes[1].legend()
axes[1].grid(True)

# Subplot 3: Quality Metric
axes[2].plot(daily_performance['date'], daily_performance['failure_rate'], label='Failure Rate', color='red')
axes[2].set_title('Daily Failure Rate Over Time')
axes[2].set_ylabel('Failure Rate (%)')
axes[2].grid(True)

plt.xlabel('Date')
plt.tight_layout()
plt.show()
```

```python
# Extract hour of day from event timestamp
df7_merged['hour'] = df7_merged['event_ts'].dt.hour

# Map station details (city, location_type, etc.) and partner/vehicle details if possible
# Merge station metadata
df7_merged = df7_merged.merge(df5[['station_id', 'city', 'location_type']], on='station_id', how='left')

# For vehicle class, we can check if it is linked to rider segment or partner. Let's merge rider details via df4 (partners) if rider has FP linkage
# First, check df4 structure and columns to link with df7
print("Columns in fleet_partners:")
print(df4.columns.tolist())

# Let's inspect the unique vehicle classes and segments represented in the swaps data
# We will write a function to group quality metrics across multiple dimensions
def analyze_failures(df, group_col):
    summary = df.groupby(group_col).agg(
        total_swaps=('event_id', 'count'),
        completed_swaps=('event_type', lambda x: (x == 'swap_completed').sum()),
        failed_no_battery=('event_type', lambda x: (x == 'failed_no_charged_battery').sum()),
        failed_system_error=('event_type', lambda x: (x == 'failed_system_error').sum()),
        abandoned_swaps=('event_type', lambda x: (x == 'abandoned_queue').sum()),
        avg_wait_time_sec=('queue_wait_sec', 'mean')
    ).reset_index()
    summary['failure_rate'] = (summary['failed_no_battery'] + summary['failed_system_error']) / summary['total_swaps']
    return summary

# Analyze by Hour of Day
hourly_summary = analyze_failures(df7_merged, 'hour')
print("\n--- Quality & Failures by Hour of Day ---")
display(hourly_summary.head(24))

# Analyze by Location Type
location_summary = analyze_failures(df7_merged, 'location_type')
print("\n--- Quality & Failures by Station Location Type ---")
display(location_summary)
```

```python
# Calculate observed discount percentage for each event in df7_merged
df7_merged['observed_discount_pct'] = np.round((df7_merged['discount_inr'] / df7_merged['list_price_inr']) * 100, 1)

# Map each PARTNER discount to its vehicle class using df4
# Construct a dictionary mapping discount_pct to vehicle_class
# Note: Where multiple partners have the same discount rate, we can map to their common vehicle class
# Let's inspect unique vehicle classes by discount level
discount_to_class = df4.groupby('discount_pct')['vehicle_class'].agg(lambda x: list(set(x))).to_dict()
print("Mapping of discount percent to vehicle classes:", discount_to_class)

# Let's create a mapping of discount percent directly to vehicle class. For 8% (CargoTuk is 3W, DabbaXpress is 2W), and 12% (ZipDrop is 2W, MetroMove is mixed)
# Let's check how many riders fall under these classes, or if we can use tariff_code == 'STD' as Standard (which are typically 2W or 3W depending on station slots)
# We can also check df7's battery_out_id prefix to differentiate 2W vs 3W directly!
df7_merged['battery_prefix_class'] = df7_merged['battery_out_id'].astype(str).apply(lambda x: '2W' if '2W' in x else ('3W' if '3W' in x else 'Unknown'))

print("\n--- Vehicle Class from Battery Out Prefix counts in Swap Events ---")
print(df7_merged['battery_prefix_class'].value_counts())

# Group quality metrics by this exact vehicle class
vehicle_summary = analyze_failures(df7_merged, 'battery_prefix_class')
print("\n--- Quality & Failures by Vehicle Class ---")
display(vehicle_summary)
```

```python
# Determine intended vehicle class from the incoming battery (battery_in_id)
df7_merged['incoming_battery_class'] = df7_merged['battery_in_id'].astype(str).apply(lambda x: '2W' if '2W' in x else ('3W' if '3W' in x else 'Unknown'))

# Let's see the distribution of incoming battery classes
print("Incoming battery class distribution:")
print(df7_merged['incoming_battery_class'].value_counts())

# Calculate quality and failures using this true class indicator
true_vehicle_summary = df7_merged.groupby('incoming_battery_class').agg(
    total_attempts=('event_id', 'count'),
    completed_swaps=('event_type', lambda x: (x == 'swap_completed').sum()),
    failed_no_battery=('event_type', lambda x: (x == 'failed_no_charged_battery').sum()),
    failed_system_error=('event_type', lambda x: (x == 'failed_system_error').sum()),
    abandoned_swaps=('event_type', lambda x: (x == 'abandoned_queue').sum()),
    avg_wait_time_sec=('queue_wait_sec', 'mean')
).reset_index()

true_vehicle_summary['failure_rate'] = (true_vehicle_summary['failed_no_battery'] + true_vehicle_summary['failed_system_error']) / true_vehicle_summary['total_attempts']
true_vehicle_summary['abandon_rate'] = true_vehicle_summary['abandoned_swaps'] / true_vehicle_summary['total_attempts']

print("\n--- True Comparative Failure Metrics by Vehicle Class ---")
display(true_vehicle_summary)
```

## Pricing and Partner Economics
Let's analyze how the pricing configurations (base pricing, partner discounts, peak/off-peak surcharges) influence swap timing, revenues, and contribution margins.

```python
import pandas as pd
import numpy as np

# 1. Compare Partner vs Standard Swaps
print("--- Overall Financial Performance by Tariff Code ---")
tariff_summary = df7_merged.groupby('tariff_code').agg(
    total_swaps=('event_id', 'count'),
    total_revenue_inr=('amount_charged_inr', 'sum'),
    total_recharge_cost_inr=('recharge_cost_inr', 'sum'),
    avg_list_price_inr=('list_price_inr', 'mean'),
    avg_discount_inr=('discount_inr', 'mean'),
    avg_amount_charged_inr=('amount_charged_inr', 'mean')
).reset_index()

tariff_summary['contribution_margin_inr'] = tariff_summary['total_revenue_inr'] - tariff_summary['total_recharge_cost_inr']
tariff_summary['margin_pct'] = (tariff_summary['contribution_margin_inr'] / tariff_summary['total_revenue_inr']) * 100
tariff_summary['avg_margin_per_swap'] = tariff_summary['contribution_margin_inr'] / tariff_summary['total_swaps']
display(tariff_summary)

# 2. Analyze Swaps by Hour of Day to see peak/off-peak timing behaviors
print("\n--- Swap Volume and Average Revenue by Hour ---")
hourly_pricing = df7_merged.groupby(['tariff_code', 'hour']).agg(
    total_swaps=('event_id', 'count'),
    avg_charged_inr=('amount_charged_inr', 'mean')
).reset_index()

# Pivot for easier side-by-side comparison
hourly_pricing_pivot = hourly_pricing.pivot(index='hour', columns='tariff_code', values=['total_swaps', 'avg_charged_inr'])
display(hourly_pricing_pivot.head(10))
```

**Root Cause Analysis of Retention**
Let's isolate new riders and construct metrics around their early-stage experiences (e.g., first-swap wait times, initial failures, ticket submittals) to identify the drivers of user churn.

```python
import pandas as pd
import numpy as np

# 1. Identify each rider's chronological swap timeline
df7_sorted = df7_merged.sort_values(by=['rider_id', 'event_ts']).copy()
df7_sorted['swap_index'] = df7_sorted.groupby('rider_id').cumcount() + 1

# Get total swaps per rider
rider_swap_counts = df7_sorted.groupby('rider_id').size().reset_index(name='total_rider_swaps')

# Define churned/one-and-done riders (who completed exactly 1 swap or had a failed first try and never returned)
# A robust metric of immediate churn is riders who made <= 2 lifetime attempts
rider_swap_counts['retained'] = rider_swap_counts['total_rider_swaps'] > 2

# 2. Extract first-experience details for each rider
first_experiences = df7_sorted[df7_sorted['swap_index'] == 1].copy()
first_experiences = first_experiences.merge(rider_swap_counts, on='rider_id', how='left')

# 3. Correlate first-experience failures/wait times with retention
print("--- Impact of First Event Type on Rider Retention ---")
event_impact = first_experiences.groupby('event_type').agg(
    total_riders=('rider_id', 'count'),
    retention_rate=('retained', 'mean'),
    avg_first_wait_sec=('queue_wait_sec', 'mean')
).reset_index()
display(event_impact)

# 4. Impact of support tickets on retention
print("\n--- Impact of Early Support Tickets on Rider Retention ---")
# Check if rider submitted any support ticket
ticket_riders = df6['rider_id'].unique()
first_experiences['submitted_ticket'] = first_experiences['rider_id'].isin(ticket_riders)

ticket_impact = first_experiences.groupby('submitted_ticket').agg(
    total_riders=('rider_id', 'count'),
    retention_rate=('retained', 'mean')
).reset_index()
display(ticket_impact)
```
